## Load the Sonar Dataset from UCI Repository

```python
def accuracy(y_pred,y_test):
 accuracy = np.mean(y_pred == y_test)
 return accuracy
acc=accuracy(y_pred,y_test_sonar)
print(f"Model Accuracy: {acc:.3f}")
```

## 1.2. SVM with GridSearchCV and 5-Fold Cross-Validation
---

## Create a Pipeline for GridSearchCV

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

## Define the Parameter Grid for Different Kernels

```python
param_grids = [
    {'svm__kernel': ['linear'], 'svm__C': [0.1, 1, 10,100]},
    {'svm__kernel': ['poly'], 'svm__C': [0.1, 1, 10],'svm__gamma':  ['auto','scale', 0.01,0.001], 'svm__degree': [2, 3,4]},
    {'svm__kernel': ['rbf'], 'svm__C': [0.1, 1, 10,100], 'svm__gamma':  ['auto','scale', 0.001,0.01]},
]
    
```

## Create Stratified 5-Fold Cross-Validation

```python
pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('svm',SVC(random_state=42))
])

grid = GridSearchCV(
    estimator=pipe,
    param_grid=param_grids,
    cv=skf,                 # Stratified 5-fold
    scoring='accuracy',    
    n_jobs=-1,              
    verbose=1               
)
```

## Initialize and Run GridSearchCV on Training Data

```python
grid.fit(X_train_sonar, y_train_sonar)
```

## Display Best Parameters and Cross-Validation Score

```python
print("\n Best Parameters:", grid.best_params_)
print(f" Best Cross Validation Score:, {grid.best_score_:.3f}")
```

## Evaluate the Best Model on the Test Set

```python
best_model = grid.best_estimator_
new_predictions = best_model.predict(X_test_sonar)
```

## Display Top 5 Parameter Combinations

```python
results = pd.DataFrame(grid.cv_results_)

cols = ['mean_test_score', 'std_test_score', 'params']
top5 = results.sort_values(by='mean_test_score', ascending=False).head(5)[cols]

print(" Top 5 parameter combinations (by mean CV accuracy):")
print(top5)
```

## Display Classification Report for Tuned SVM

```python
print("\nClassification Report:\n", classification_report(y_test_sonar, new_predictions, target_names=["Mine", "Rock"]))
```

## Display Confusion Matrix for Tuned SVM

```python
 confusion=confusion_matrix(y_test_sonar,new_predictions)
print(confusion)
```

## 2.1. Multinomial Logistic Regression
---

## Define the Multinomial Logistic Regression Class

```python
import numpy as np

class MultinomialLogisticRegression:
    def __init__(self, learning_rate=0.01, epochs=1000, reg_lambda=0.0):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.reg_lambda = reg_lambda
        self.weights = None
        self.bias = None
        self.classes_ = None

    # Softmax 
    def _softmax(self, z):
        z = z - np.max(z, axis=1, keepdims=True) #for overflow of large numbers
        exp_z = np.exp(z)
        return exp_z / np.sum(exp_z, axis=1, keepdims=True)

    #  Cross-Entropy Loss
    def _cross_entropy(self, y_true, y_pred):
        m = y_true.shape[0]
        loss = -np.sum(y_true * np.log(y_pred + 1e-15)) / m
        reg = (self.reg_lambda / (2 * m)) * np.sum(self.weights ** 2)
        return loss + reg

    #  Fit 
    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.classes_ = np.unique(y)
        n_classes = len(self.classes_)

        # one-hot encoding
        Y = np.zeros((n_samples, n_classes))
        for i, label in enumerate(y):
            Y[i, np.where(self.classes_ == label)[0][0]] = 1

     
        self.weights = np.zeros((n_classes, n_features))
        self.bias = np.zeros((n_classes, 1))

        for epoch in range(self.epochs):
          
            z = np.dot(X, self.weights.T) + self.bias.T
            probs = self._softmax(z)

            # gradient descent algorithmb
            error = probs - Y
            dw = (np.dot(error.T, X) / n_samples) + (self.reg_lambda / n_samples) * self.weights
            db = np.mean(error, axis=0).reshape(n_classes, 1)

            
            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db


    # Predict Probabilities with softmax function
    def predict_proba(self, X):
        z = np.dot(X, self.weights.T) + self.bias.T
        return self._softmax(z)

    # Predict Labels
    def predict(self, X):
        probs = self.predict_proba(X)
        indices = np.argmax(probs, axis=1)
        return self.classes_[indices]

    #  Accuracy Score
    def score(self, X, y):
        y_pred = self.predict(X)
        return np.mean(y_pred == y)

    #  For GridSearchCV Compatibility
    def get_params(self, deep=True):
        return {
            "learning_rate": self.learning_rate,
            "epochs": self.epochs,
            "reg_lambda": self.reg_lambda
        }

    def set_params(self, **params):
        for key, value in params.items():
            setattr(self, key, value)
        return self
            
              
          
       
          

       
        
                
          
          
            
        
    
```

## Define Hyperparameter Grid for Multinomial Logistic Regression

```python
param_grid = {
    "learning_rate": [0.01, 0.05, 0.1],
    "epochs": [200, 300, 500],
    "reg_lambda": [0.0, 0.01, 0.1]
}
```

## Run GridSearchCV for Multinomial Logistic Regression

```python
model = MultinomialLogisticRegression()


grid = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=5,               
    scoring='accuracy', 
    n_jobs=-1,        
    verbose=2
)


grid.fit(X_train, y_train_beans)
```

###  Describe how a decision tree builds its decision structure. (5 points)

```python
'''A decision tree builds its structure using a top-down, greedy algorithm. 
Starting from the full dataset, 
it evaluates all features and possible split thresholds to select the split that yields the largest reduction in impurity (measured by Gini or entropy). 
It recursively repeats this process on each resulting subset until a stopping condition is met. 
The final leaf nodes represent class predictions, forming decision regions in the feature space'''
```
